налаштування бібліотек

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

налаштування стилю графіків

```python
sns.set_theme(style="whitegrid")
```

Візуалізація даних
Датасет: Auto MPG (UCI).
Вимоги: Multivariate, Mixed attributes, Missing values.
ми бачемо 5 рядків а на справді їх 398 ми взяли завдяки url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/auto-mpg/auto-mpg.data"
cols = ['mpg', 'cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year', 'origin', 'name']

df = pd.read_csv(url, names=cols, na_values='?', comment='\t', sep=r'\s+', skipinitialspace=True)
df.head()
```


Пошук та заповнення пропущених значень.

```python
print(df.isnull().sum())
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mean())
print("\nПісля очистки:", df.isnull().sum().sum())
```

Залежність атрибутів (Scatter plot)
Зв'язок потужності (horsepower) та ваги (weight).

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='horsepower', y='weight', hue='cylinders')
plt.show()
```

Гістограма (5 діапазонів)
Розподіл показника MPG.

```python
plt.figure(figsize=(8, 5))
plt.hist(df['mpg'], bins=5, color='teal', edgecolor='black')
plt.show()
```

Багатовимірна візуалізація
MPG vs Displacement з урахуванням регіону (origin) та прискорення (size).

```python
plt.figure(figsize=(12, 7))
sns.scatterplot(data=df, x='displacement', y='mpg', hue='origin', size='acceleration', alpha=0.7)
plt.legend(bbox_to_anchor=(1.05, 1))
plt.show()
```

 Додаткові графіки (Розподіл та кореляція)
 Графік 6: Boxplot
 Графік 7: Heatmap
 Графік 8: Violin plot

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='origin', y='mpg', data=df)
plt.show()

plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()

plt.figure(figsize=(8, 5))
sns.violinplot(x='cylinders', y='horsepower', data=df)
plt.show()
```
